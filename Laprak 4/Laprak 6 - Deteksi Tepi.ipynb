## Import Library

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
#202331175_Rizalwirap
```

## Membaca Gambar

```python
image = cv.imread("parkir.jpg")
#202331175_Rizalwirap
```

## Menampilkan Gambar

## Mendeteksi Tepi Pada Gambar

```python
image.shape
#202331175_Rizalwirap
```

```python
gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
edges = cv.Canny(gray, 75, 150)
#202331175_Rizalwirap
```

```python
cv.imshow("Gambar Parkir", edges)
cv.waitKey(0)
cv.destroyAllWindows()
#202331175_Rizalwirap
```

```python
fig, axs = plt.subplots(1,3, figsize = (10,10))
ax = axs.ravel()

axs[0].imshow(image, cmap = "gray")
axs[0].set_title("Parkir 3 channel RGB")

axs[1].imshow(gray, cmap = "gray")
axs[1].set_title("Parkir convert Grayscale")

axs[2].imshow(edges, cmap = "gray")
axs[2].set_title("Parkir tepi Canny")
#202331175_Rizalwirap
```

## Deteksi Garis Lurus

```python
lines = cv.HoughLinesP(edges, 1, np.pi/180, 30, maxLineGap = 250)
image_line = image.copy()
#202331175_Rizalwirap
```

```python
for line in lines:
    x1, y1, x2, y2 = line[0]
    cv.line(image_line, (x1,y1), (x2,y2), (100,78,95), 1)

#202331175_Rizalwirap
```

```python
fig, axs = plt.subplots(1,2, figsize = (10,10))
ax = axs.ravel()

axs[0].imshow(edges, cmap = "gray")
axs[0].set_title("Deteksi Tepi")

axs[1].imshow(image_line, cmap = "gray")
axs[1].set_title("Deteksi Garis Lurus")
#202331175_Rizalwirap
```
