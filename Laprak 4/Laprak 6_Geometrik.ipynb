```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
#202331175_Rizalwirap
```

```python
plat_bgr = cv.imread('plat.jpg')

plat_rgb = cv.cvtColor(plat_bgr, cv.COLOR_BGR2RGB)

plt.imshow(plat_rgb)
plt.show()
#202331175_Rizalwirap
```

```python
from skimage import io, transform #202331175_Rizalwirap
```

```python
src = np.array([
    [0,0],
    [0,50],
    [300,50],
    [300,0]
])

crop = np.array([
    [373,191],
    [346,257],
    [516,273],
    [542,195],
])

tform = transform.ProjectiveTransform()
tform.estimate(src,crop)

warped = transform.warp(plat_rgb, tform, output_shape=(50,300))

fig, axs = plt.subplots(1,2, figsize = (10,5))
ax = axs.ravel()

axs[0].imshow(warped)
axs[0].set_title("Gambar Img 1")

axs[1].imshow(plat_rgb)
axs[1].plot(crop[:,0], crop[:,1], '.r')
axs[1].set_title("Gambar Asli")

for a in axs:
    a.axis('Off')

plt.tight_layout()
plt.show #202331175_Rizalwirap
```
