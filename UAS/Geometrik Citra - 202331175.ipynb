```python
#Rizal Wira Pambudi 202331175
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Rizal Wira Pambudi 202331175
image = cv2.imread('Rizal Close.jpeg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
```

```python
#Rizal Wira Pambudi 202331175
image.shape
```

```python
original = image.copy()
```

```python
#Rizal Wira Pambudi 202331175

#ambil height width
(h, w) = image.shape[:2]
#menemukan titik tengah gambar
center = (w // 2, h // 2)
#rotasi berdasarkan titik tengah
M_rotate = cv2.getRotationMatrix2D(center, 45, 1.0)
#rotasi berdasarkan height dan width
rotated = cv2.warpAffine(image, M_rotate, (w, h))
# ubah ukuran gambar
resized = cv2.resize(image, (60, 120))
# edit crop gambar
cropped = image[755:3845, 440:2745]
# Membalikkan Gambar
flipped = cv2.flip(image, 1)
# translasi ukuran gambar
M_translate = np.float32([[1, 0, 100], [0, 1, 60]])
translated = cv2.warpAffine(image, M_translate, (w, h))
```

## Menampilkan citra hasil transformasi geometri

```python
#Rizal Wira Pambudi 202331175
fig, axs = plt.subplots(2, 3, figsize=(12, 8))

axs[0, 0].imshow(original)
axs[0, 0].set_title("Citra Asli")
axs[0, 0].axis('on')

axs[0, 1].imshow(rotated)
axs[0, 1].set_title("After Rotated")
axs[0, 1].axis('on')

axs[0, 2].imshow(resized)
axs[0, 2].set_title("After Resized")
axs[0, 2].axis('on')

axs[1, 0].imshow(cropped)
axs[1, 0].set_title("After Cropped")
axs[1, 0].axis('on')

axs[1, 1].imshow(flipped)
axs[1, 1].set_title("After Flipped")
axs[1, 1].axis('on')

axs[1, 2].imshow(translated)
axs[1, 2].set_title("After Translated")
axs[1, 2].axis('on')

plt.tight_layout()
plt.show()
```
