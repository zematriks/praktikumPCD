```python
#Rizal Wira Pambudi 202331175
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os
import io
```

```python
#Rizal Wira Pambudi 202331175 - load gambar
original_image = cv2.imread("orikompres.jpeg")
```

```python
#Rizal Wira Pambudi 202331175 - konversi rgb
gambar_rgb = cv2.cvtColor(original_image, cv2.COLOR_BGR2RGB)
```

```python
#Rizal Wira PAmbudi 202331175
jpeg_quality = 10
encode_param = [int(cv2.IMWRITE_JPEG_QUALITY), jpeg_quality]
result, encimg = cv2.imencode('.jpg', original_image, encode_param)

gambar_lossy = cv2.imdecode(encimg, 1)
gambar_lossy_rgb = cv2.cvtColor(gambar_lossy, cv2.COLOR_BGR2RGB)

levels = 4
divider = 256 // levels

hasil_kuantisasi = (gambar_rgb // divider) * divider + (divider // 2)

# Pastikan tipe datanya tetap uint8 (integer 0-255)
hasil_kuantisasi = np.uint8(hasil_kuantisasi)

print("Kompresi gambar lossy dan kuantisasi warna telah selesai!")
```

```python
#Rizal Wira PAmbudi 202331175
ukuran_original = original_image.nbytes / 1024

ukuran_lossy = len(encimg) / 1024

ukuran_kuantisasi = hasil_kuantisasi.nbytes / 1024

plt.figure(figsize=(20, 10))

#Gambar Original
plt.subplot(1, 3, 1)
plt.imshow(gambar_rgb)

plt.title(f'Gambar Original\nUkuran: {ukuran_original:.2f} KB (In-Memory)', fontsize=14)
plt.axis('off')

# GAmbar Lossy
plt.subplot(1, 3, 2)
plt.imshow(gambar_lossy_rgb)
plt.title(f'Gambar Lossy JPEG Q={jpeg_quality}\nUkuran: {ukuran_lossy:.2f} KB', fontsize=14)
plt.axis('off')

#Gambar Kuantisasi Warna
plt.subplot(1, 3, 3)
plt.imshow(hasil_kuantisasi)
plt.title(f'Hasil Kuantisasi RGB ({levels} Level)\nUkuran: {ukuran_kuantisasi:.2f} KB (In-Memory)', fontsize=14)
plt.axis('off')

plt.tight_layout(rect=[0, 0, 1, 0.95]) 
plt.show()
```
